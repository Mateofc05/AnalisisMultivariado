```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn import metrics

import statsmodels.api as sm
import statsmodels.stats.api as sms
from statsmodels.stats.diagnostic import het_breuschpagan, het_white
from statsmodels.stats.outliers_influence import variance_inflation_factor, OLSInfluence

%matplotlib inline
```

```python
# Datos Real Estate
df = pd.read_csv('Realestate.csv')

# Renombrar columnas
df.columns = ['No', 'Fecha_Transaccion', 'Antiguedad', 'Distancia_MRT', 
              'Tiendas_Conveniencia', 'Latitud', 'Longitud', 'Precio']

print("Datos cargados exitosamente")
print(f"Forma del dataset: {df.shape}")
df.head()
```

**Respuestas – No-linealidad**

1. La variable que muestra la relación no-lineal más clara con el Precio es la **Distancia al MRT**: el precio baja muy rápido en los primeros ~1000 metros y después casi se aplana, formando una curva parecida a una exponencial decreciente. La **Antigüedad** también es no-lineal, porque tiene una forma de "U": las casas nuevas y las más viejas (más de 35 años) tienen precios un poco más altos, mientras que las de entre 15 y 25 años son las más baratas. En cambio, el número de **Tiendas de Conveniencia** se ve más o menos lineal y creciente, aunque como es una variable discreta los puntos se agrupan en columnas.

   - **Tiendas de Conveniencia:** dejarla lineal, ya que su relación con el precio es aproximadamente recta.
   - **Precio (variable respuesta):** se podría probar con `log(Precio)`, porque tiene sesgo a la derecha y un valor muy alto (117.5) que se sale del resto de los datos.


```python
# Comparar modelos con y sin transformación
X = df.drop(['Precio', 'No'], axis=1)
y = df['Precio']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Modelo lineal con Distancia MRT
modelo_lineal = LinearRegression()
modelo_lineal.fit(X_train[['Distancia_MRT']], y_train)
y_pred_lineal = modelo_lineal.predict(X_test[['Distancia_MRT']])

# Modelo logarítmico con Distancia MRT
X_train_log = np.log(X_train[['Distancia_MRT']])
X_test_log = np.log(X_test[['Distancia_MRT']])

modelo_log = LinearRegression()
modelo_log.fit(X_train_log, y_train)
y_pred_log = modelo_log.predict(X_test_log)

# Comparar R²
print("=== COMPARACIÓN DE MODELOS ===")
print(f"Modelo Lineal (Distancia MRT):")
print(f"  R² = {metrics.r2_score(y_test, y_pred_lineal):.4f}")
print(f"  RMSE = {np.sqrt(metrics.mean_squared_error(y_test, y_pred_lineal)):.4f}")
print(f"\nModelo Logarítmico (log Distancia MRT):")
print(f"  R² = {metrics.r2_score(y_test, y_pred_log):.4f}")
print(f"  RMSE = {np.sqrt(metrics.mean_squared_error(y_test, y_pred_log)):.4f}")
```

```python
# Visualizar la mejora de la transformación
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Modelo lineal
axes[0].scatter(X_test['Distancia_MRT'], y_test, alpha=0.4, s=15, label='Datos reales')
axes[0].scatter(X_test['Distancia_MRT'], y_pred_lineal, alpha=0.4, s=15, color='red', label='Predicción')
axes[0].set_xlabel('Distancia MRT')
axes[0].set_ylabel('Precio')
axes[0].set_title(f'Modelo Lineal - R² = {metrics.r2_score(y_test, y_pred_lineal):.4f}')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Modelo logarítmico
axes[1].scatter(X_test_log, y_test, alpha=0.4, s=15, label='Datos reales')
axes[1].scatter(X_test_log, y_pred_log, alpha=0.4, s=15, color='green', label='Predicción')
axes[1].set_xlabel('log(Distancia MRT)')
axes[1].set_ylabel('Precio')
axes[1].set_title(f'Modelo Logarítmico - R² = {metrics.r2_score(y_test, y_pred_log):.4f}')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
#  Regresión polinómica de diferentes grados
print("=== REGRESIÓN POLINÓMICA ===")

grados = [1, 2, 3, 4]
resultados = []

for grado in grados:
    # Crear pipeline con transformación polinómica
    modelo_poly = make_pipeline(
        PolynomialFeatures(degree=grado, include_bias=False),
        LinearRegression()
    )
    
    # Entrenar
    modelo_poly.fit(X_train[['Distancia_MRT']], y_train)
    y_pred_poly = modelo_poly.predict(X_test[['Distancia_MRT']])
    
    # Métricas
    r2 = metrics.r2_score(y_test, y_pred_poly)
    rmse = np.sqrt(metrics.mean_squared_error(y_test, y_pred_poly))
    aic = len(y_test) * np.log(metrics.mean_squared_error(y_test, y_pred_poly)) + 2 * (grado + 1)
    
    resultados.append({
        'Grado': grado,
        'R²': r2,
        'RMSE': rmse,
        'AIC': aic
    })

df_resultados = pd.DataFrame(resultados)
print(df_resultados.round(4))
```

```python
# Visualizar regresión polinómica
plt.figure(figsize=(12, 5))

# Gráfico de datos
plt.subplot(1, 2, 1)
plt.scatter(X_test['Distancia_MRT'], y_test, alpha=0.4, s=15, label='Datos Real E.')

# Crear rango para predicciones suaves
X_range = np.linspace(X_test['Distancia_MRT'].min(), X_test['Distancia_MRT'].max(), 100).reshape(-1, 1)

for grado, color in zip([1, 2, 3, 4], ['blue', 'green', 'orange', 'red']):
    modelo_poly = make_pipeline(
        PolynomialFeatures(degree=grado, include_bias=False),
        LinearRegression()
    )
    modelo_poly.fit(X_train[['Distancia_MRT']], y_train)
    y_range = modelo_poly.predict(X_range)
    plt.plot(X_range, y_range, color=color, linewidth=2, label=f'Grado {grado}')

plt.xlabel('Distancia MRT')
plt.ylabel('Precio')
plt.title('Regresión Polinómica - Diferentes Grados')
plt.legend()
plt.grid(True, alpha=0.3)

# Gráfico de R² vs Grado
plt.subplot(1, 2, 2)
plt.plot(df_resultados['Grado'], df_resultados['R²'], 'o-', color='blue', linewidth=2)
plt.xlabel('Grado del Polinomio')
plt.ylabel('R²')
plt.title('R² vs Grado del Polinomio')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
#  Modelo completo y análisis de residuos
X_const = sm.add_constant(X)
modelo_completo = sm.OLS(y, X_const).fit()
print(modelo_completo.summary())
```

**Respuesta:**

No se observa un patrón claro en forma de embudo. En la gráfica de residuos vs predichos, la dispersión de los residuos es más o menos la misma a lo largo de todos los valores predichos, y en la gráfica de escala-localización (√|Residuos|) los puntos no muestran una tendencia a subir o bajar. Lo que sí se alcanza a notar es:
- Para valores predichos bajos (menores a 17) casi todos los residuos son positivos, y entre 30 y 40 la mayoría son negativos. Esto parece más una señal de **no-linealidad** (por ejemplo, por usar la distancia sin log) que de heteroscedasticidad.
- Hay algunos puntos muy alejados, como uno con residuo cercano a 75, que son posibles **outliers**.


```python
# Prueba de Breusch-Pagan
print("=== PRUEBA DE BREUSCH-PAGAN ===\n")

bp_test = het_breuschpagan(modelo_completo.resid, X_const)
print(f"LM Statistic: {bp_test[0]:.4f}")
print(f"LM p-value: {bp_test[1]:.4f}")
print(f"F Statistic: {bp_test[2]:.4f}")
print(f"F p-value: {bp_test[3]:.4f}")

if bp_test[1] < 0.05:
    print("\n❌ Rechazamos H₀: Hay evidencia de heteroscedasticidad")
else:
    print("\n✅ No rechazamos H₀: No hay evidencia suficiente de heteroscedasticidad")
```

```python
# Prueba de White
print("=== PRUEBA DE WHITE ===\n")

white_test = het_white(modelo_completo.resid, X_const)
print(f"LM Statistic: {white_test[0]:.4f}")
print(f"LM p-value: {white_test[1]:.4f}")
print(f"F Statistic: {white_test[2]:.4f}")
print(f"F p-value: {white_test[3]:.4f}")

if white_test[1] < 0.05:
    print("\n❌ Rechazamos H₀: Hay evidencia de heteroscedasticidad")
else:
    print("\n✅ No rechazamos H₀: No hay evidencia suficiente de heteroscedasticidad")
```

**Respuestas – Heteroscedasticidad**

1. En la prueba de **Breusch-Pagan** el p-value es **0.2064** y en la prueba de **White** es **0.6249**. Como ambos son mayores a 0.05, **no rechazamos H₀**, así que no hay evidencia suficiente de heteroscedasticidad; es decir, podemos suponer que la varianza de los errores es constante. (La prueba de White muestra una advertencia de matriz singular porque al elevar al cuadrado y multiplicar variables como Latitud, Longitud y Fecha aparecen columnas casi repetidas, pero el resultado apunta a la misma conclusión que Breusch-Pagan).

2. Sí, los resultados son consistentes con el análisis visual, ya que en las gráficas de residuos no se observó un patrón de embudo.

3. Para la inferencia esto quiere decir que los errores estándar de MCO, las pruebas t y los intervalos de confianza del modelo son **confiables** en cuanto a la varianza, y no es estrictamente necesario corregirlos. Sin embargo, hay que tener cuidado porque en el resumen del modelo la prueba de Jarque-Bera indica que los residuos **no son normales** (curtosis de 16.9), probablemente por los outliers, y el número de condición es muy grande, lo que indica multicolinealidad. Por eso los p-values deben interpretarse con algo de precaución.


```python
# Errores estándar robustos (Sandwich)
print("=== SOLUCIÓN 1: ERRORES ESTÁNDAR ROBUSTOS ===\n")

# Modelo con errores robustos HC3
modelo_robusto = sm.OLS(y, X_const).fit(cov_type='HC3')

print("Comparación de errores estándar:")
print(f"{'Variable':<20} {'MCO':<15} {'Robusto':<15} {'Cambio %':<10}")
print("-" * 60)

for var in X_const.columns:
    se_mco = modelo_completo.bse[var]
    se_rob = modelo_robusto.bse[var]
    cambio = (se_rob - se_mco) / se_mco * 100
    print(f"{var:<20} {se_mco:<15.4f} {se_rob:<15.4f} {cambio:>+8.2f}%")

print("\n PREGUNTA: ¿Cómo cambian los errores estándar?")

# Escribe tu respuesta aquí
```

**Respuesta:**

Los coeficientes no cambian, solo cambian los errores estándar, y los cambios son moderados (entre −17% y +23%). Los errores estándar **aumentan** en Tiendas de Conveniencia (+23.13%), Distancia MRT (+16.36%), Antigüedad (+11.73%) y Latitud (+5.00%), mientras que **disminuyen** en la constante (−17.38%), Longitud (−9.23%) y Fecha de Transacción (−5.62%). A pesar de estos cambios, las variables que eran significativas lo siguen siendo; por ejemplo, Tiendas de Conveniencia pasa de un t de 6.02 a aproximadamente 4.89. Como las pruebas no detectaron heteroscedasticidad, estas diferencias probablemente se deben más a la influencia de algunos outliers que a una varianza no constante.


```python
#  Mínimos Cuadrados Ponderados (WLS)
print("=== SOLUCIÓN 2: MÍNIMOS CUADRADOS PONDERADOS (WLS) ===\n")

# Estimar la varianza usando los residuos al cuadrado
# Modelo auxiliar: log(e²) ~ X
residuos_sq = modelo_completo.resid ** 2
log_residuos_sq = np.log(residuos_sq + 1e-10)  # Evitar log(0)

modelo_aux = sm.OLS(log_residuos_sq, X_const).fit()
var_predicha = np.exp(modelo_aux.fittedvalues)

# Pesos = 1/varianza
pesos = 1 / var_predicha

# Modelo WLS
modelo_wls = sm.WLS(y, X_const, weights=pesos).fit()

print("Comparación de coeficientes:")
print(f"{'Variable':<20} {'MCO':<15} {'WLS':<15} {'Cambio %':<10}")
print("-" * 60)

for var in X_const.columns:
    coef_mco = modelo_completo.params[var]
    coef_wls = modelo_wls.params[var]
    cambio = (coef_wls - coef_mco) / abs(coef_mco) * 100
    print(f"{var:<20} {coef_mco:<15.4f} {coef_wls:<15.4f} {cambio:>+8.2f}%")

print("\n PREGUNTA: ¿Cómo cambian los coeficientes estimados?")

# Escribe tu respuesta aquí
```

**Respuesta:**

Con WLS los coeficientes sí cambian, aunque todos conservan el mismo signo:
- **Tiendas de Conveniencia** aumenta de 1.13 a 1.39 (+22.93%).
- **Antigüedad** se vuelve un poco más negativa, de −0.270 a −0.303.
- **Distancia MRT** se vuelve menos negativa, de −0.0045 a −0.0038.
- **Fecha de Transacción** baja de 5.15 a 4.40 (−14.59%).
- **Latitud** casi no cambia (+0.78%).
- **Longitud** tiene el cambio porcentual más grande (−142.73%), pero esto no es tan importante porque esa variable **no es significativa** (p = 0.798) y su error estándar es enorme, así que su coeficiente es muy inestable.

En general, WLS le da menos peso a las observaciones con residuos grandes, por lo que los coeficientes se ajustan un poco hacia el comportamiento de la mayoría de los datos.


```python
# Comparación de modelos con diferentes soluciones
print("=== COMPARACIÓN DE SOLUCIONES ===\n")

print("Modelo MCO (sin corrección):")
print(f"  R² = {modelo_completo.rsquared:.4f}")
print(f"  AIC = {modelo_completo.aic:.2f}")
print(f"  BIC = {modelo_completo.bic:.2f}")

print("\nModelo Robusto (HC3):")
print(f"  R² = {modelo_robusto.rsquared:.4f}")
print(f"  AIC = {modelo_robusto.aic:.2f}")
print(f"  BIC = {modelo_robusto.bic:.2f}")

print("\nModelo WLS:")
print(f"  R² = {modelo_wls.rsquared:.4f}")
print(f"  AIC = {modelo_wls.aic:.2f}")
print(f"  BIC = {modelo_wls.bic:.2f}")
```

**Respuestas – Soluciones**

1. Yo prefiero el **modelo MCO con errores estándar robustos (HC3)**. Como las pruebas de Breusch-Pagan y White no encontraron heteroscedasticidad, no es necesario cambiar la forma de estimar los coeficientes, y los errores robustos funcionan como un "seguro" que no afecta las estimaciones. Aunque WLS tiene un R² más alto (**0.7060** contra 0.5824) y un AIC más bajo (2919.51 contra 2987.92), estas métricas no se pueden comparar directamente, porque en WLS se calculan sobre los datos ponderados y no sobre los datos originales. Además, los pesos se estimaron con los residuos de MCO, que incluyen outliers, lo que puede hacer que algunos pesos no sean muy confiables.

2. No, los errores robustos **no cambian los coeficientes**. Esto es porque HC3 solo cambia la forma de calcular la matriz de varianzas de los coeficientes (es decir, sus errores estándar), pero los coeficientes se siguen calculando con la misma fórmula de MCO. Por eso el R², el AIC y el BIC son exactamente iguales en el modelo MCO y en el robusto.

3. La ventaja de WLS es que, si el modelo de la varianza es correcto, produce estimaciones **más eficientes**, o sea, con menor varianza. WLS sí corrige el problema desde la estimación, dándole menos peso a las observaciones con más ruido, mientras que los errores robustos solo corrigen la inferencia (errores estándar, pruebas t e intervalos) sin mejorar los coeficientes. La desventaja es que WLS depende de modelar bien la varianza; si los pesos están mal estimados, los resultados pueden ser peores.


```python
#  Detección de outliers
print("=== DETECCIÓN DE OUTLIERS ===\n")

# Calcular influencia
influencia = OLSInfluence(modelo_completo)

# Distancia de Cook
cook_dist = influencia.cooks_distance[0]
umbral_cook = 4 / len(y)

# Leverage
leverage = influencia.hat_matrix_diag
umbral_leverage = 2 * X_const.shape[1] / len(y)

# Residuos estandarizados
residuos_est = influencia.resid_studentized_internal

# Identificar outliers
outliers_cook = np.where(cook_dist > umbral_cook)[0]
outliers_leverage = np.where(leverage > umbral_leverage)[0]
outliers_residuos = np.where(np.abs(residuos_est) > 3)[0]

print(f"Distancia de Cook > {umbral_cook:.4f}: {len(outliers_cook)} observaciones")
print(f"Leverage > {umbral_leverage:.4f}: {len(outliers_leverage)} observaciones")
print(f"|Residuos estandarizados| > 3: {len(outliers_residuos)} observaciones")
```

```python
#  RANSAC
print("=== RANSAC ===\n")

from sklearn.linear_model import RANSACRegressor

# Modelo RANSAC
ransac = RANSACRegressor(
    estimator=LinearRegression(),
    min_samples=0.5,
    residual_threshold=5.0,
    random_state=42
)
ransac.fit(X_train, y_train)

# Predicciones
y_pred_ransac = ransac.predict(X_test)

print(f"Número de inliers: {ransac.inlier_mask_.sum()}")
print(f"Número de outliers: {(~ransac.inlier_mask_).sum()}")

print("\nMétricas en test:")
print(f"RANSAC: R² = {metrics.r2_score(y_test, y_pred_ransac):.4f}, RMSE = {np.sqrt(metrics.mean_squared_error(y_test, y_pred_ransac)):.4f}")
```
